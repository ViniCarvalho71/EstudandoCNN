```python
model = models.Sequential()

model.add(layers.Conv2D(32,(5,5),activation='relu', input_shape=(28,28,1)))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Conv2D(64, (5, 5), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))

model.add(layers.Flatten())
model.add(layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
from keras.datasets import mnist
from keras.utils import to_categorical


(train_images, train_labels), (test_images, test_labels) = mnist.load_data()  
                              
    
train_images = train_images.reshape((60000, 28, 28, 1))
train_images = train_images.astype('float32') / 255


test_images = test_images.reshape((10000, 28, 28, 1))
test_images = test_images.astype('float32') / 255


train_labels = to_categorical(train_labels)
test_labels = to_categorical(test_labels)


batch_size = 100
epochs = 10


model.compile(loss='categorical_crossentropy',
              optimizer='sgd',
              metrics=['accuracy'])



model.fit(train_images, train_labels,
          batch_size=batch_size,
          epochs=epochs,
          verbose=1)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)

print('Test loss:', test_loss)
print('Test accuracy:', test_acc)
```

```python
import numpy as np
from pathlib import Path
from keras import Input, Model

# ============================================================
# CONFIGURAÇÃO
# ============================================================

OUTPUT_DIR = Path("pesos_fpga")
OUTPUT_DIR.mkdir(exist_ok=True)

MULT_SHIFT = 16  # bits de deslocamento fixos para todo o projeto


# ============================================================
# QUANTIZAÇÃO INT8 DOS PESOS
# ============================================================

def quantize_int8(values):
    """
    Quantização simétrica:
        INT8 = [-127, 127]
    Retorna: (quantized, scale)
    """
    max_value = np.max(np.abs(values))
    scale = max_value / 127.0 if max_value != 0 else 1.0
    quantized = np.round(values / scale)
    quantized = np.clip(quantized, -127, 127)
    return quantized.astype(np.int8), scale


def save_mem(filename, values):
    """Salva valores INT8 em hexadecimal de 8 bits (complemento de 2)."""
    values = np.asarray(values).flatten()
    with open(filename, "w") as f:
        for value in values:
            value_uint8 = np.uint8(value)
            f.write(f"{value_uint8:02X}\n")


def save_scale(filename, scale):
    with open(filename, "w") as f:
        f.write(f"{scale:.12f}\n")


# ============================================================
# CONV1 (índice 0 no model.add(...))
# ============================================================

weights_conv1, bias_conv1 = model.layers[0].get_weights()
quant_conv1, scale_conv1 = quantize_int8(weights_conv1)
quant_bias_conv1, scale_bias_conv1 = quantize_int8(bias_conv1)

save_mem(OUTPUT_DIR / "conv1_weights.mem", quant_conv1)
save_mem(OUTPUT_DIR / "conv1_bias.mem", quant_bias_conv1)
save_scale(OUTPUT_DIR / "conv1_scale.txt", scale_conv1)
save_scale(OUTPUT_DIR / "conv1_bias_scale.txt", scale_bias_conv1)

print("Conv1  -> scale pesos:", scale_conv1, " scale bias:", scale_bias_conv1)

# ============================================================
# CONV2 (índice 2 no model.add(...))
# ============================================================

weights_conv2, bias_conv2 = model.layers[2].get_weights()
quant_conv2, scale_conv2 = quantize_int8(weights_conv2)
quant_bias_conv2, scale_bias_conv2 = quantize_int8(bias_conv2)

save_mem(OUTPUT_DIR / "conv2_weights.mem", quant_conv2)
save_mem(OUTPUT_DIR / "conv2_bias.mem", quant_bias_conv2)
save_scale(OUTPUT_DIR / "conv2_scale.txt", scale_conv2)
save_scale(OUTPUT_DIR / "conv2_bias_scale.txt", scale_bias_conv2)

print("Conv2  -> scale pesos:", scale_conv2, " scale bias:", scale_bias_conv2)

# ============================================================
# DENSE (índice 5 no model.add(...): 0=Conv1,1=Pool1,2=Conv2,3=Pool2,4=Flatten,5=Dense)
# ============================================================

weights_dense, bias_dense = model.layers[5].get_weights()
quant_dense, scale_dense = quantize_int8(weights_dense)
quant_bias_dense, scale_bias_dense = quantize_int8(bias_dense)

save_mem(OUTPUT_DIR / "dense_weights.mem", quant_dense)
save_mem(OUTPUT_DIR / "dense_bias.mem", quant_bias_dense)
save_scale(OUTPUT_DIR / "dense_scale.txt", scale_dense)
save_scale(OUTPUT_DIR / "dense_bias_scale.txt", scale_bias_dense)

print("Dense  -> scale pesos:", scale_dense, " scale bias:", scale_bias_dense)


# ============================================================
# MULTIPLICADOR EM PONTO FIXO (inteiro + shift, sem float em VHDL)
# ============================================================
# acc_int32 * scale_real  ~=  (acc_int32 * M) >> SHIFT
# Mesma conta que a requant_unit vai fazer em hardware.

def quantize_multiplier(real_multiplier, shift=MULT_SHIFT):
    multiplier_int = int(round(real_multiplier * (1 << shift)))
    return multiplier_int, shift


def requantize_int32_to_int8(acc, multiplier_int, shift=MULT_SHIFT):
    """Simula exatamente a conta da requant_unit em VHDL. Use para
    gerar valores de referência dos testbenches."""
    product = int(acc) * int(multiplier_int)
    rounding = 1 << (shift - 1)
    result = (product + rounding) >> shift
    return int(np.clip(result, -127, 127))


# ============================================================
# CALIBRAÇÃO: ESCALA DE SAÍDA (ATIVAÇÃO) DE CADA CAMADA
# ============================================================
# MaxPool não precisa de calibração própria (usa a escala da
# convolução anterior, já que só escolhe um valor existente).

def calibrate_activation_scale(model, layer_index, images, num_samples=200):
    """Escala (max_abs/127) da saída da camada `layer_index`,
    calculada rodando `num_samples` imagens de exemplo pelo modelo."""
    inp = Input(shape=images.shape[1:])
    x = inp
    for layer in model.layers[:layer_index + 1]:
        x = layer(x)
    intermediate_model = Model(inputs=inp, outputs=x)

    sample = images[:num_samples]
    activations = intermediate_model.predict(sample, verbose=0)

    max_abs = np.max(np.abs(activations))
    return max_abs / 127.0 if max_abs > 0 else 1.0


scale_input = 1.0 / 127.0  # pixels normalizados em [0,1] tratados como INT8

# --- Conv1 ---
scale_conv1_out = calibrate_activation_scale(model, layer_index=0, images=train_images)
mult_conv1_real = (scale_input * scale_conv1) / scale_conv1_out
mult_conv1_int, shift_conv1 = quantize_multiplier(mult_conv1_real)

# Pool1 usa a MESMA escala de saída do Conv1

# --- Conv2 ---
scale_conv2_out = calibrate_activation_scale(model, layer_index=2, images=train_images)
mult_conv2_real = (scale_conv1_out * scale_conv2) / scale_conv2_out
mult_conv2_int, shift_conv2 = quantize_multiplier(mult_conv2_real)

# Pool2 usa a MESMA escala de saída do Conv2

# --- Dense (final) ---
# Só necessário se quiser uma saída INT8 pra depuração. Se for usar
# argmax direto no acumulador INT32, pode ignorar mult_dense_*.
scale_dense_out = calibrate_activation_scale(model, layer_index=5, images=train_images)
mult_dense_real = (scale_conv2_out * scale_dense) / scale_dense_out
mult_dense_int, shift_dense = quantize_multiplier(mult_dense_real)


# ============================================================
# SALVAR MULTIPLICADORES
# ============================================================

def save_multiplier(filename, multiplier_int, shift):
    with open(filename, "w") as f:
        f.write(f"MULTIPLIER={multiplier_int}\n")
        f.write(f"SHIFT={shift}\n")


save_multiplier(OUTPUT_DIR / "conv1_multiplier.txt", mult_conv1_int, shift_conv1)
save_multiplier(OUTPUT_DIR / "conv2_multiplier.txt", mult_conv2_int, shift_conv2)
save_multiplier(OUTPUT_DIR / "dense_multiplier.txt", mult_dense_int, shift_dense)

print("\nMultiplicadores (para constantes/generics em VHDL):")
print("Conv1  -> M:", mult_conv1_int, " shift:", shift_conv1)
print("Conv2  -> M:", mult_conv2_int, " shift:", shift_conv2)
print("Dense  -> M:", mult_dense_int, " shift:", shift_dense)

print("\nArquivos gerados em", OUTPUT_DIR.resolve(), ":")
for file in sorted(OUTPUT_DIR.iterdir()):
    print(" -", file.name)
```
